```python
import pandas as pd, matplotlib.pyplot as plt
w = pd.read_csv("data/raw/wearable.csv", parse_dates=["timestamp"])
p = w[w.patient_id == 1].head(96 * 3)      # first 3 days
p.plot(x="timestamp", y="glucose", figsize=(12, 4))
plt.axhline(180, color="r", linestyle="--")  # spike threshold
plt.show()
print((w.glucose > 180).mean())  # share of readings above 180
```

```python
import pandas as pd, matplotlib.pyplot as plt
cgm = pd.read_csv("data/processed/cgmacros_all.csv.gz", parse_dates=["time"])
p = cgm[cgm.subject == 3].set_index("time")
ax = p["libre"].iloc[:1440 * 3].plot(figsize=(12, 4))
meals = p[p.carbs.notna()].iloc[:12]
ax.scatter(meals.index, meals["libre"], color="orange", zorder=3, label="meal")
ax.axhline(180, color="r", linestyle="--")
ax.legend(); plt.show()
```

```python
import pandas as pd
cgm = pd.read_csv("data/processed/cgmacros_all.csv.gz")
miss = cgm.groupby("subject")[["hr", "mets", "act_kcal"]].agg(lambda s: s.isna().mean())
print(miss[miss.hr > 0.5].round(2))                       # people with mostly-missing heart rate
print(cgm.groupby("subject").group.first().value_counts())  # people per group in the raw data

feats = pd.read_csv("data/processed/features.csv.gz")
print(feats.groupby("group").subject.nunique())            # people per group after features
```

```python
import pandas as pd
f = pd.read_csv("data/processed/features.csv.gz")
f = f[f.glu_now <= 180]
print(f[f.group == "healthy"].groupby("subject").apply(lambda d: (d.future_max > 180).mean()).round(3).sort_values(ascending=False).head(5))
```
